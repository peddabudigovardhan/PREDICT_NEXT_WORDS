```python
quotes=df['quote']
```

```python
quotes=quotes.str.lower()
```

```python
import string
translator=str.maketrans('','',string.punctuation)
quotes=quotes.apply(lambda x:x.translate(translator))
```

```python
quotes.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
from IPython.utils.sysinfo import num_cpus
vocab_size=10000
tokenizer=Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index=tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequence=tokenizer.texts_to_sequences(quotes)
```

```python
print(sequence[:4])
```

```python
X=[]
y=[]
for seq in sequence:
  for i in range(1,len(seq)):
    input_seq=seq[:i]
    output_seq=seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
X
```

```python
len(X)
```

```python
len(y)
```

```python
max_len=max(len(x) for x in X)
print(max_len)
```

```python

from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padding=pad_sequences(X,maxlen=max_len,padding='pre')
```

```python

X_padding
```

```python
X_padding[0]
```

```python
y=np.array(y)
```

```python
X_padding.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot=to_categorical(y,num_classes=vocab_size)
```

```python
y_one_hot[0]
```

```python
y_one_hot.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,SimpleRNN,Dense
```

```python
embid_dim=50
rnn_units=128
```

```python
rnn_model=Sequential()
rnn_model.add(
    Embedding(input_dim=vocab_size,input_length=max_len,output_dim=embid_dim)
)
rnn_model.add(
    SimpleRNN(units=rnn_units,return_sequences=False)
)
rnn_model.add(
    Dense(units=vocab_size,activation='softmax')
)
```

```python
rnn_model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
rnn_model.summary()
```

```python
lstm_model=Sequential()
lstm_model.add(
    Embedding(input_dim=vocab_size,input_length=max_len,output_dim=embid_dim)
)
lstm_model.add(
    LSTM(units=rnn_units)
)
lstm_model.add(
    Dense(units=vocab_size,activation='softmax')
)
```

```python
lstm_model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```
